```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import shap

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import warnings
warnings.filterwarnings("ignore")
from google.colab import files

uploaded = files.upload()
import zipfile

with zipfile.ZipFile("archive.zip", "r") as zip_ref:
    zip_ref.extractall("dataset")

print("Dataset extracted successfully!")
import os

print(os.listdir("dataset"))
# Load the Dataset

df = pd.read_csv("dataset/india_2000_2024_daily_weather.csv")

# Display first 5 rows
df.head()
# Basic Information

print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
# Dataset Information

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
# Check for Duplicate Rows

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
# Remove Duplicate Rows

df.drop_duplicates(inplace=True)

print("New Shape:", df.shape)
# Convert Date Column

df["date"] = pd.to_datetime(df["date"])

print(df["date"].head())
# Extract Year, Month and Day

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day

df.head()
# Drop Original Date Column

df.drop("date", axis=1, inplace=True)

df.head()
# Check Correlation Between Rain Variables

print(df[["precipitation_sum", "rain_sum"]].corr())
# Remove Redundant Feature
df.drop("precipitation_sum", axis=1, inplace=True)

print("Remaining Columns:")
print(df.columns)
# Encode City Names

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["city"] = le.fit_transform(df["city"])

df.head()
# Final Dataset Check

df.info()

display(df.head())
# Distribution of Daily Rainfall

plt.figure(figsize=(8,5))

sns.histplot(df["rain_sum"], bins=50, kde=True)

plt.xlim(0, 50)   # Focus on the most common rainfall range

plt.title("Distribution of Daily Rainfall")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")

plt.show()
# Box Plot of Rainfall

plt.figure(figsize=(8,4))

sns.boxplot(x=df["rain_sum"])

plt.title("Box Plot of Daily Rainfall")

plt.show()
# Average Rainfall by City

city_names = le.inverse_transform(df["city"].unique())

city_rain = df.groupby("city")["rain_sum"].mean().reset_index()

city_rain["city"] = le.inverse_transform(city_rain["city"])

plt.figure(figsize=(12,6))

sns.barplot(data=city_rain, x="city", y="rain_sum")

plt.xticks(rotation=45)

plt.title("Average Rainfall by City")

plt.xlabel("City")

plt.ylabel("Average Rainfall (mm)")

plt.show()
# Correlation Heatmap

plt.figure(figsize=(10,8))

sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
# Define Features and Target

X = df.drop("rain_sum", axis=1)

y = df["rain_sum"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

# Split Dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)

# Train Random Forest Regressor

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model trained successfully!")
# Make Predictions

y_pred = model.predict(X_test)

print("Predictions generated successfully!")
# Model Evaluation

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Model Performance")
print("-" * 30)
print(f"Mean Absolute Error (MAE): {mae:.3f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.3f}")
print(f"R² Score: {r2:.3f}")
# Actual vs Predicted Rainfall

plt.figure(figsize=(8,6))

plt.scatter(y_test, y_pred, alpha=0.5)

plt.xlabel("Actual Rainfall (mm)")
plt.ylabel("Predicted Rainfall (mm)")
plt.title("Actual vs Predicted Rainfall")

plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         color="red",
         linestyle="--")

plt.show()
# Feature Importance

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

plt.figure(figsize=(10,6))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance")

plt.show()

display(feature_importance)

# SHAP Explainability

# Use a representative sample of the test data for faster SHAP analysis
X_shap = X_test.sample(n=100, random_state=42)

# Create SHAP TreeExplainer for the trained Random Forest
explainer = shap.TreeExplainer(model)

# Calculate SHAP values for the sampled test data
shap_values = explainer.shap_values(X_shap)


# ---------------------------------------------------------
# 1. SHAP Summary Plot
# ---------------------------------------------------------

shap.summary_plot(
    shap_values,
    X_shap,
    show=False
)

plt.title("SHAP Summary Plot")
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 2. SHAP Feature Importance
# ---------------------------------------------------------

shap.summary_plot(
    shap_values,
    X_shap,
    plot_type="bar",
    show=False
)

plt.title("SHAP Feature Importance")
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 3. SHAP Waterfall Plot - Individual Prediction
# ---------------------------------------------------------
# Select the sampled observation with the highest predicted rainfall
sample_positions = X_test.index.get_indexer(X_shap.index)
sample_predictions = y_pred[sample_positions]

sample_index = int(np.argmax(sample_predictions))

# Get the original row index from X_test
original_index = X_shap.index[sample_index]

# Get the corresponding actual and predicted rainfall values
actual_value = y_test.loc[original_index]
prediction_position = X_test.index.get_loc(original_index)
predicted_value = y_pred[prediction_position]

print(f"Actual Rainfall: {actual_value:.2f}")
print(f"Predicted Rainfall: {predicted_value:.2f}")

base_value = float(
    np.asarray(explainer.expected_value).reshape(-1)[0]
)

sample_shap_values = np.asarray(
    shap_values[sample_index]
).reshape(-1)

# Validate SHAP explanation
shap_prediction = base_value + sample_shap_values.sum()

print(f"SHAP reconstructed prediction: {shap_prediction:.6f}")
print(f"Model prediction: {predicted_value:.6f}")
print(f"Difference: {abs(shap_prediction - predicted_value):.10f}")

sample_explanation = shap.Explanation(
    values=sample_shap_values,
    base_values=base_value,
    data=X_shap.iloc[sample_index].values,
    feature_names=X_shap.columns.tolist()
)

shap.plots.waterfall(
    sample_explanation,
    max_display=len(X_shap.columns),
    show=False
)

plt.title("SHAP Waterfall Plot - Individual Rainfall Prediction")
plt.tight_layout()
plt.show()

# Compare Actual and Predicted Values

comparison = pd.DataFrame({
    "Actual Rainfall": y_test.values,
    "Predicted Rainfall": y_pred
})

comparison.head(10)
comparison.to_csv("prediction_results.csv", index=False)

print("Prediction results saved successfully!")
```
